```python
import sqlite3
import pandas as pd
from pathlib import Path

DB_PATH = Path("data/europe_trade_network.db")
```

```python
conn = sqlite3.connect(DB_PATH)

# 1) List tables
tables = pd.read_sql_query("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'table'
    ORDER BY name
""", conn)

print(f"Database: {DB_PATH}")
print(f"Tables found: {len(tables)}")
print(tables)
```

```python
# 2) Show schema + row count + sample rows for each table
for table_name in tables["name"]:
    print("\n" + "=" * 80)
    print(f"TABLE: {table_name}")

    schema = pd.read_sql_query(f"PRAGMA table_info('{table_name}')", conn)
    print("\nSchema:")
    print(schema[["cid", "name", "type", "notnull", "pk"]])

    row_count = pd.read_sql_query(f"SELECT COUNT(*) AS n_rows FROM '{table_name}'", conn)
    print("\nRow count:")
    print(row_count)

    sample = pd.read_sql_query(f"SELECT * FROM '{table_name}' LIMIT 5", conn)
    print("\nSample rows:")
    print(sample)

conn.close()
print("\nDone.")
```
